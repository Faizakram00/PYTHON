```python
# some set operation

# union: combine element from two sets excluding duplicate

# intersection: only common element betwwn sets

# difference: return the elements that is present in first set and not in second set

# symmetric difference: returns element that are present in either of sets but not in both
```

```python
s1 = {"hiking","reading","coding"}
s2 = {"coding","travelling","ph0tography"}

```

```python
# intersection 
s1 & s2
```

```python
# union 
s1 | s2
```

```python
# difference
s1-s2
```

```python
# symmetric difference
s1 ^ s2
```
