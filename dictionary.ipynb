```python
# dictionary comprehension
students = ["Faiz","Kaif","Saif"]
marks = [80,95,90]

student_marks = {}
for student,mark in zip(students,marks):
    student_marks[student] = marks

print(student_marks)
```

```python
# Another method
students = ["Faiz","Kaif","Saif"]
marks = [80,95,90]

for i in zip(students,marks):
    print(i)
```

```python
#dictionary comprehension
students = ["Faiz","Kaif","Saif"]
marks = [80,95,90]
{students:marks for students,marks in zip(students,marks)}
```

```python
user_id = [1,2,3]
pswd = ["12m","fm12","lv02"]

{user_id:pswd for user_id,pswd in zip(user_id,pswd)}
```
