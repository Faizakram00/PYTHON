# if-elif-else

```python
a = 10
if a > 100:
    print("will execute if a is greater than 100")
elif a < 100:
    print("will execute if a is lesser than 100")
else:
    print("number is equal")
```

```python
score = 20
if score >=90:
    print("grade A")
elif score >=75:
    print("Grade B")
elif score >= 65:
    print("Grade c")
elif score >= 45:
    print("Grade D")
elif score >= 33:
    print("Grade E")
else:
    print("fail")
```

```python
# Nested if else condition

x = 1
y = 10

if x > 5:
    if y > 5:
        print("Both are greater than 5")
    else:
        print("x is greater  than 5")

else:
    print("x is not greater than 5")
```
