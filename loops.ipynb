# loops allow you to repeatedly execute a block of code
#  Two types of loops. 1. while   2. for loop

# while >> repeatedly execute a block of code as long as condition true.

```python
n = 7
i  = 1
while i < n:
    print(i)
    i = i+1
```

```python
count = 5
while count > 0:
    print(count)
    count = count - 1
```

```python
n = 7
i = 1
while i < n:
    print(i)
    i = i + 1
else:
    print("This will be executed when the while loop is run successfully")
```

```python
n = 7
i = 1
while i < n:
    print(i)
    i = i + 1
    if i == 3:
        break
else:
    print("This will be executed when the while loop is run successfully")
```

```python
n = 7
i = 1
while n > i:
    i = i+1
    if i == 3:
        continue
    print(i)
else:
    print("thiw will be executed the while loop is run successfully")
```
